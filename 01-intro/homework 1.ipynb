```python
#Q3 Fuel types
Fuel = df['fuel_type'].nunique()
print(f"Fuel types in the dataset: {Fuel}")
```

```python
#Q4 Missing values
df.isna().sum()
```

```python
#Q5 Max fuel efficiency
asian_cars = df[df['origin'] == 'Asia']
Max_fuel_efficiency = asian_cars['fuel_efficiency_mpg'].max()
print(f"Max fuel efficiency of Asian cars: {Max_fuel_efficiency}")
```

```python
#Q6 Median value of horsepower
horse_power = df['horsepower'].median()
print(f"Median value of horsepower: {horse_power}")

most_frequent = df['horsepower'].mode()[0]
print(f"Most frequent value of horsepower: {most_frequent}")

df_filled = df['horsepower'].fillna(most_frequent)

filled_horse_power = df_filled.median()
print(f"Median value of horsepower after filling missing values: {filled_horse_power}")

```

```python
#Q7 Sum of weights

#1.Select all the cars from Asia
asian_cars = df[df['origin'] == 'Asia']
#2.Select only columns vehicle_weight and model_year
df_sub = asian_cars[['vehicle_weight', 'model_year']]
#3 Select the first 7 values
df_first_7 = df_sub.head(7)
#4 Get the underlying NumPy array. Let's call it X.
X = df_first_7.to_numpy()
print(f"Underlying NumPy array (X):\n{X}\n")
#5 Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
XTX = X.T @ X
print(f"X transpose multiplied by X (XTX):\n{XTX}\n")
#6 invert XTX
XTX_inv = np.linalg.inv(XTX)
print(f"inverse of XTX (XTX_inv):\n{XTX_inv}")
#7 Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
y=np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print(f"Array y:\n{y}\n\n")
#8 Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
w_temp = XTX_inv @ X.T
w = w_temp @ y
print(f"Weights (w):\n{w}")

total_sum = w.sum()
print(f"Sum of weights: {total_sum}")
```
